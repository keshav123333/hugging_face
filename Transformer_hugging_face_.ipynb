# Categorization

if jana ki ye topic kis label ke andar ya topic ke andar aate

````python
from transformers import pipeline

# Load zero-shot classification pipeline
catego = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli",   # specify model
    model_kwargs={"force_download": True}
)

# Example texts
texts = ["BIO tech journal", "nano tech journal", "land dispute"]

# Candidate labels
labels = ['science', 'law', 'land_law']

# Run classification for each text
for t in texts:
    result = catego(t, candidate_labels=labels)
    print(f"\nText: {t}")
    print(result)
````

## 🚀 2️⃣ Using a Specific Model

Agar tumhe ek particular model chahiye, toh `model=` parameter se specify karo.

### 🔹 GPT-2 (default small)

```python
generator = pipeline("text-generation", model="gpt2")
```

### 🔹 GPT-Neo (open-source alternative to GPT-3)

```python
generator = pipeline("text-generation", model="EleutherAI/gpt-neo-2.7B")
```

### 🔹 GPT-J (more powerful, creative writing / coding ke liye)

```python
generator = pipeline("text-generation", model="EleutherAI/gpt-j-6B")
```

### 🔹 Falcon (modern lightweight LLM)

```python
generator = pipeline("text-generation", model="tiiuae/falcon-7b-instruct")
```

---

## 💬 3️⃣ Control Output — Parameters

You can control how the text is generated:

```python
text = "AI will change the future because"

result = generator(
    text,
    max_new_tokens=50,     # how much to generate
    temperature=0.7,       # creativity (0 = deterministic, 1+ = more random)
    top_p=0.9,             # nucleus sampling — limits to top 90% probability mass
    do_sample=True,        # enable randomness
    num_return_sequences=2 # generate multiple variations
)

for r in result:
    print(r["generated_text"])
```

## 🧩 4️⃣ Use Cases by Model Type

| Use Case                      | Recommended Model                    | Notes                          |
| ----------------------------- | ------------------------------------ | ------------------------------ |
| Story / blog generation       | `EleutherAI/gpt-j-6B`                | More coherent long text        |
| Chat-style replies            | `tiiuae/falcon-7b-instruct`          | Instruction-tuned              |
| Coding / technical text       | `codellama/CodeLlama-7b-Instruct-hf` | For code generation            |
| Short completions             | `gpt2`                               | Lightweight and fast           |
| Research / scientific writing | `mistralai/Mistral-7B-Instruct-v0.1` | High-quality instruction model |

## 💡 Summary (yaad rakhne layak)

| Concept                       | Meaning                    |
| ----------------------------- | -------------------------- |
| `pipeline("text-generation")` | Creates text generator     |
| `model="..."`                 | Chooses which model to use |
| `temperature`                 | Controls randomness        |
| `max_new_tokens`              | How much text to generate  |
| `top_p`, `do_sample`          | Sampling strategies        |
